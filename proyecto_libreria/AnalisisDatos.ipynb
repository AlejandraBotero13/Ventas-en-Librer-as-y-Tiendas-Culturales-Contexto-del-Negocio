# Taller Pandas — Ventas Librerías Culturales

```python
import pandas as pd
import numpy as np
```

## 1.Lectura y Verificación de Importación

```python
#Carga el archivo ventas_libreria_cultural.csv en un DataFrame denominado df_libros evitando la adición de columnas índices innecesarias.
df_libros = pd.read_csv("ventas_libreria_cultural.csv", index_col=False)
df_libros
```

## 2. Selección Unidimensional y Multivariable:

```python
#Extrae la columna Genero_Libro como una Serie de Pandas y genera una nueva tabla que contenga simultáneamente Sucursal, 
# Formato y Precio_Lista.
genero = df_libros['Genero_Libro']
print(type(genero))
tabla = df_libros[['Sucursal', 'Formato', 'Precio_Lista']]
tabla.head(2)
```

## 3.Acceso por Etiqueta (loc)

```python
# Consulta todos los campos de la transacción registrada exactamente en la fila con índice 310 utilizando el método .
df_libros.loc[310]
```

## 4. Tratamiento de Redundancias

```python
#Aplica .drop_duplicates() para suprimir las filas repetidas y confirmar el tamaño limpio de la base de datos.
print('Antes:', df_libros.shape)
df_libros = df_libros.drop_duplicates()
print('Después:', df_libros.shape)
```

## 5. Filtrado por Volumen de Compra

```python
#Realiza un filtro booleano simple que seleccione únicamente las compras mayoristas donde Unidades_Vendidas > 3.
df_libros[df_libros['Unidades_Vendidas'] > 3]
df_libros.head(13)
```

## 6. Filtrado Categórico con .isin()

```python
#Recupera todas las operaciones registradas exclusivamente en las sucursales Campus Universitario y Galería de Arte.
df_libros[df_libros['Sucursal'].isin(['Campus Universitario', 'Galería de Arte'])].head()
```

## 7. Filtrado por Rango de Precios (between)

```python
#Utiliza .between() para aislar las compras donde el Precio_Lista se ubique entre $45,000 y $75,000 inclusive.
df_libros[df_libros['Precio_Lista'].between(45000, 75000)].head(10)
```

## 8. Consultas Directas con .query()

```python
#Realiza una búsqueda mediante .query() para extraer las ventas efectuadas
# con Metodo_Pago == 'Transferencia QR' en las que Unidades_Vendidas >= 2.
df_libros.query("Metodo_Pago == 'Transferencia QR' and Unidades_Vendidas >= 2").head(10)
```

## 9. Filtrado Compuesto con Métodos de Texto

```python
#Construye una condición booleana con & que localice los libros cuyo Genero_Libro inicie con "Ensayo" mediante .str.startswith() 
# y cuyo Precio_Lista sea superior a $50,000.
filtro = df_libros['Genero_Libro'].str.startswith('Ensayo') & (df_libros['Precio_Lista'] > 50000)
df_libros[filtro].head(10)
```

## 10. Imputación Estratificada de Faltantes

```python
#Rellena los valores ausentes (NaN) en la columna Precio_Lista calculando el promedio aritmético de dicha variable, e imputa los nulos
#  de Descuento_Aplicado con un valor constante de 0.0 mediante .fillna().
df_libros = df_libros.fillna({
    'Precio_Lista': df_libros['Precio_Lista'].mean(),
    'Descuento_Aplicado': 0.0
})
df_libros.isna().sum()
```

## 11. Transformación con Funciones y Creación de Columnas (apply)

```python
#Define una función de Python que calcule el subtotal neto a pagar: (Precio_Lista * (1 - Descuento_Aplicado)) * Unidades_Vendidas. Luego, 
# aplícala con .apply() para crear la columna Total_Venta.
def calcular_total(fila):
    return (fila['Precio_Lista'] * (1 - fila['Descuento_Aplicado'])) * fila['Unidades_Vendidas']

df_libros['Total_Venta'] = df_libros.apply(calcular_total, axis=1)
df_libros.head()
```

## 12. Renombrado y Reorganización de Variables

```python
#Renombra Genero_Libro a Categoria y Descuento_Aplicado a Descuento usando .rename(). Posteriormente, reordena
#  las columnas colocando ID_Transaccion y Total_Venta al inicio de la estructura.
df_libros = df_libros.rename(columns={'Genero_Libro': 'Categoria', 'Descuento_Aplicado': 'Descuento'})
primeras = ['ID_Transaccion', 'Total_Venta']
df_libros = df_libros[primeras + [c for c in df_libros.columns if c not in primeras]]
df_libros.head()
```

## 13. Agrupación y Métricas de Facturación por Sucursal

```python
# Agrupa los registros por Sucursal y calcula simultáneamente el promedio de Precio_Lista y
#  la suma global acumulada de Total_Venta mediante el método .agg().
df_libros.groupby('Sucursal').agg({'Precio_Lista': 'mean', 'Total_Venta': 'sum'}).round(1)
```

## 14. Métricas de Rango Personalizadas

```python
# Diseña una función personalizada llamada rango (series.max() - series.min()) y
#  aplícala con .agg() sobre las columnas Precio_Lista y Total_Venta tras agrupar por Formato.
def rango(series):
    return series.max() - series.min()

df_libros.groupby('Formato').agg({'Precio_Lista': rango, 'Total_Venta': rango})
```

## 15. Agrupamiento multinivel

```python
#Realiza una agrupación combinada por Sucursal y Categoria, obteniendo un resumen que detalle el
# promedio del Total_Venta y la suma global de Unidades_Vendidas para cada combinación.
df_libros.groupby(['Sucursal', 'Categoria']).agg({'Total_Venta': 'mean', 'Unidades_Vendidas': 'sum'}).round(2).head(100)
```
